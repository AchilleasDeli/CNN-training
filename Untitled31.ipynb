```python
# Import necessary libraries
import tensorflow as tf
import matplotlib.pyplot as plt
import numpy as np

# Load Fashion-MNIST dataset which is already saved in tenserflow notebook
fashion_mnist = tf.keras.datasets.fashion_mnist
(x_train, y_train), (x_test, y_test) = fashion_mnist.load_data()

# create class names (could not do this as well)
class_names = ['T-shirt/top', 'Trouser', 'Pullover', 'Dress', 'Coat',
               'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle boot']

# Normalize pixel values (we can have max of 255 so divide it to normalize between 1-0
x_train, x_test = x_train / 255.0, x_test / 255.0

# For fun display some images 
def plot_sample_images():
    plt.figure(figsize=(10, 5))
    for i in range(10):
        plt.subplot(2, 5, i + 1)
        plt.xticks([])
        plt.yticks([])
        plt.grid(False)
        plt.imshow(x_train[i], cmap=plt.cm.binary)
        plt.xlabel(class_names[y_train[i]])
    plt.show()

# Plot sample images
plot_sample_images()

# Print dataset shapes
print(f"Training set: {x_train.shape}, Labels: {y_train.shape}")
print(f"Test set: {x_test.shape}, Labels: {y_test.shape}")
```

```python
#We can see we have a train set of 60.000 and a test set of 10000 
```

```python
#PART A 
#We basically create a CNN which firstly the Conv2D layer applies 32 filters of size (5x5) to detect patterns 
#like edges, textures, and shapes.
#It uses a pretty agressive activation function which is the relu function.
#Then to further downscale the size we take a max polling , which takes a 2x2 region scans the image and keeps the strongest activations 
#Then The second Conv2D layer uses 64 filters of size (5x5) to learn more advanced patterns.
#Another MaxPooling layer downsamples again, making the network focus on the most important features.
#Then we flatten to create a 1 dimensional vector since a nn excpects a 1D vector as input
#then we create a fully connected layer with 1,024 neurons.
#then we dropout the 20% of neurons to avoid overfitting 
#It receives a 1D input vector (after flattening the CNN's feature maps).
#Each neuron is connected to every neuron in the next layer
#The Softmax activation converts the output into probabilities (0 to 1), summing to 1.
```

```python
# Define CNN Model 
model = tf.keras.models.Sequential([
    tf.keras.layers.Conv2D(32, kernel_size=(5, 5), input_shape=(28, 28, 1), activation='relu'),
    tf.keras.layers.MaxPool2D(pool_size=(2, 2)),
    tf.keras.layers.Conv2D(64, kernel_size=(5, 5), activation='relu'),
    tf.keras.layers.MaxPool2D(pool_size=(2, 2)),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(1024, activation='relu'),
    tf.keras.layers.Dropout(0.2),
    tf.keras.layers.Dense(10, activation='softmax')
])

# Compile model
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# Train and Evaluate for 5, 10, 20 epochs
epochs_list = [5, 10, 20]
results = {}

for epochs in epochs_list:
    print(f"\nTraining for {epochs} epochs...")
    model.fit(x_train, y_train, epochs=epochs, verbose=1, validation_data=(x_test, y_test))
    res = model.evaluate(x_test, y_test, verbose=0)
    results[epochs] = res[1]  # Store accuracy

# Print accuracy results
for epochs, acc in results.items():
    print(f"Test Accuracy after {epochs} epochs: {acc:.4f}")
```

```python
# Define CNN Model with an extra convolutional + max-pooling layer
model_extra = tf.keras.models.Sequential([
    tf.keras.layers.Conv2D(32, kernel_size=(5, 5), input_shape=(28, 28, 1), activation='relu'),
    tf.keras.layers.MaxPool2D(pool_size=(2, 2)),
    tf.keras.layers.Conv2D(64, kernel_size=(5, 5), activation='relu'),
    tf.keras.layers.MaxPool2D(pool_size=(2, 2)),
    tf.keras.layers.Conv2D(128, kernel_size=(3, 3), activation='relu'),  # Added Layer
    tf.keras.layers.MaxPool2D(pool_size=(2, 2)),  # Added Layer
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(1024, activation='relu'),
    tf.keras.layers.Dropout(0.2),
    tf.keras.layers.Dense(10, activation='softmax')
])

# Compile model
model_extra.compile(optimizer='adam',
                    loss='sparse_categorical_crossentropy',
                    metrics=['accuracy'])

# Train and Evaluate for 20 epochs
print("\nTraining with Extra Conv Layer for 20 epochs...")
model_extra.fit(x_train, y_train, epochs=20, verbose=1, validation_data=(x_test, y_test))
res_extra = model_extra.evaluate(x_test, y_test, verbose=0)

# Print accuracy result
print(f"Test Accuracy after 20 epochs (with extra Conv layer): {res_extra[1]:.4f}")
```
